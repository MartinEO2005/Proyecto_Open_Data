```python
# Conteo total de ceros y conteo por columna
import pandas as pd
data = df

total_cells = data.shape[0] * data.shape[1]
zero_counts = 0
rows = []

for c in data.columns:
    ser = data[c]
    if pd.api.types.is_numeric_dtype(ser):
        z = (ser == 0).sum()
    else:
        z = ser.astype(str).str.strip().isin(["0", "0.0"]).sum()
    pct_rows = (z / len(data)) * 100
    rows.append((c, int(z), round(pct_rows, 4)))
    zero_counts += int(z)

zero_pct_total = (zero_counts / total_cells) * 100 if total_cells > 0 else 0.0

zero_df = pd.DataFrame(rows, columns=['column', 'zero_count', 'zero_pct_rows']).sort_values('zero_count', ascending=False).reset_index(drop=True)

print(f"Total celdas: {total_cells:,}")
print(f"Celdas con valor 0: {zero_counts:,}  ({zero_pct_total:.4f} % del dataset total)")
display(zero_df.head(11)) # solo las primeras 11 tienen 0s y para no hacer tan largo el output
```

Una vez obtenido los porcentajes de 0s y nulos en el dataset hay que ver si se pueden eliminar (dropear) o si, se pueden inferir de alguna manera.

1. Aunque son valores que se pueden inferir, ya que el dataset cuenta con un porcentaje muy bajo de nulos (0.8%) lo dropeo.
2. Todas las variables que contienen un porcentaje de 0s tienen sentido ya que o son booleanas o tienen posibilidad de serlo.

```python
# Creación de un nuevo DataFrame sin valores NA
df_clean = df.copy()
df_clean = df_clean.dropna(axis=0, how='any').reset_index(drop=True)
print("Shape original df:", df.shape)
print("Shape df_clean (sin NA):", df_clean.shape)
```

```python
cancelled_flights = df[df['Cancelled'] == 1]
print("Número de vuelos cancelados:", len(cancelled_flights))
```

Elimino columnas que no son necesarias, como las de fecha que son todas derivables de una misma. Las elimino para no sobrecargar tanto a la hora de manejar datos ya que son demasiadas columnas.

```python
# Eliminar columnas derivables de Date
cols_to_drop = ['Year','Quarter','Month','DayofMonth','DayOfWeek']
existing = [c for c in cols_to_drop if c in df_clean.columns]
print("Columnas a eliminar (existentes):", existing)
df_clean = df_clean.drop(columns=existing, errors='ignore')
print("Shape tras eliminar columnas derivables:", df_clean.shape)
```

```python
# filtrar por distancia irreal, manteniendo cancelados
threshold = 3  # millas
before = df_clean.shape[0]

# Normalizar nombre de columna Cancelled si hay variantes
cancel_col = None
for name in ['CANCELLED','Cancelled','Cancelled','Cancelled']:
    if name in df_clean.columns:
        cancel_col = name
        break

if 'Distance' not in df_clean.columns:
    print("Atención: no existe la columna 'Distance' en df_clean.")
else:
    # crear máscara: True para filas a mantener
    if cancel_col:
        keep_mask = (df_clean['Distance'] > threshold) | (df_clean[cancel_col] == 1)
    else:
        keep_mask = (df_clean['Distance'] > threshold)
    df_clean = df_clean[keep_mask].reset_index(drop=True)
    after = df_clean.shape[0]
    print(f"Filas antes: {before}, Filas después del filtro de distancia (umbral={threshold}): {after}, Filas eliminadas: {before-after}")
```

Normalizo el formato HHMM para que se pueda intepretar correctamente los Strings y no confundirme con minutos, horas, o lo que fuese.


```python
# Normalizar columnas de tiempo a formato HHMM (strings de 4 dígitos)
cols_time = ['CRSDepTime','CRSArrTime','DepTime','ArrTime']

for c in cols_time:
    if c in df_clean.columns:
        # Convertir a string, quitar espacios, rellenar con ceros a la izquierda
        df_clean[c] = df_clean[c].astype(str).str.strip().str.zfill(4)
        # Reemplazar valores 'NaN' que se convirtieron en string por np.nan
        df_clean[c] = df_clean[c].replace({'NaN': pd.NA, 'nan': pd.NA})
        
print("Ejemplo de valores normalizados:")
for c in cols_time:
    if c in df_clean.columns:
        print(c, "->", df_clean[c].dropna().unique()[:10])
```

```python
import pandas as pd

# Detectar nombre de columna de fecha
fecha_col = 'FlightDate'
 
# Convertir columna de fecha a datetime
df_clean[fecha_col] = pd.to_datetime(df_clean[fecha_col], errors='coerce')

# Función para limpiar HHMM
def clean_hhmm(series):
    return (
        series.astype(str)
        .str.replace(r'\.0$', '', regex=True)   # quitar ".0"
        .str.replace(r'\D', '', regex=True)     # quitar caracteres no numéricos
        .str.zfill(4)                           # rellenar a 4 dígitos
        .replace({'NaN': pd.NA, 'nan': pd.NA})
    )

# Normalizar columnas de tiempo
for c in ['CRSDepTime','CRSArrTime','DepTime','ArrTime']:
    if c in df_clean.columns:
        df_clean[c] = clean_hhmm(df_clean[c])

# Crear datetimes programados
if 'CRSDepTime' in df_clean.columns:
    hhmm = df_clean['CRSDepTime'].dropna()
    df_clean.loc[hhmm.index, 'sched_dep_dt'] = pd.to_datetime(df_clean.loc[hhmm.index, fecha_col]) + \
        pd.to_timedelta(hhmm.str[:2].astype(int), unit='h') + \
        pd.to_timedelta(hhmm.str[2:].astype(int), unit='m')

if 'CRSArrTime' in df_clean.columns:
    hhmm = df_clean['CRSArrTime'].dropna()
    df_clean.loc[hhmm.index, 'sched_arr_dt'] = pd.to_datetime(df_clean.loc[hhmm.index, fecha_col]) + \
        pd.to_timedelta(hhmm.str[:2].astype(int), unit='h') + \
        pd.to_timedelta(hhmm.str[2:].astype(int), unit='m')

# Crear datetimes reales (solo si existen valores)
if 'DepTime' in df_clean.columns:
    hhmm = df_clean['DepTime'].dropna()
    df_clean.loc[hhmm.index, 'dep_dt'] = pd.to_datetime(df_clean.loc[hhmm.index, fecha_col]) + \
        pd.to_timedelta(hhmm.str[:2].astype(int), unit='h') + \
        pd.to_timedelta(hhmm.str[2:].astype(int), unit='m')

if 'ArrTime' in df_clean.columns:
    hhmm = df_clean['ArrTime'].dropna()
    df_clean.loc[hhmm.index, 'arr_dt'] = pd.to_datetime(df_clean.loc[hhmm.index, fecha_col]) + \
        pd.to_timedelta(hhmm.str[:2].astype(int), unit='h') + \
        pd.to_timedelta(hhmm.str[2:].astype(int), unit='m')

print(df_clean[['CRSDepTime','sched_dep_dt','DepTime','dep_dt']].head(10))
```

Elimino columnas de tiempos originales porque ya cree las variables datetime y no necesito duplicados. Ademas que evidentemente, el tener tantos datos y columnas consume mucho tiempo y recursos. FlightDate la mantengo porque es mas compacto para graficas/analisis que no son por hora,etc.

```python
df_clean = df_clean.drop(columns=['CRSDepTime','CRSArrTime','DepTime','ArrTime'], errors='ignore')
# Reorganizar columnas: poner FlightDate primero, luego las nuevas datetime, luego el resto
cols_base = list(df_clean.columns)
fecha_col = 'FlightDate'  
cols_new = ['sched_dep_dt','sched_arr_dt','dep_dt','arr_dt']
ordered_cols = [fecha_col] + [c for c in cols_new if c in cols_base] + [c for c in cols_base if c not in ([fecha_col] + cols_new)]
df_clean = df_clean[ordered_cols]
df_clean.head()
```

**Normalizacion de IATAs**

La normalizacion se aplica a las columnas Origin, Dest, IATA_Code_Operating_Airline e IATA_Code_Marketing_Airline porque todas contienen codigos IATA que deben estar en un formato uniforme. 

El proceso convierte los valores a tipo string, elimina espacios en blanco, pasa todo a mayusculas y quita caracteres no alfabeticos. Esto asegura que los codigos de aeropuertos y aerolineas se comparen de forma consistente, evita duplicados por diferencias de formato y permite hacer merges y agrupaciones sin errores.

```python
# Normalizar codigos IATA de aeropuertos y aerolineas
for c in ['Origin','Dest','IATA_Code_Operating_Airline','IATA_Code_Marketing_Airline']:
    if c in df_clean.columns:
        df_clean[c] = (
            df_clean[c]
            .astype(str)              
            .str.strip()              
            .str.upper()              
            .str.replace(r'[^A-Z]', '', regex=True)
        )

# Comprobar ejemplos de valores unicos
for c in ['Origin','Dest','IATA_Code_Operating_Airline','IATA_Code_Marketing_Airline']:
    if c in df_clean.columns:
        print(f"Valores unicos en {c}:", df_clean[c].unique()[:20])
```

```python
for c in ['Cancelled','Diverted']:
    if c in df_clean.columns:
        df_clean[c] = df_clean[c].astype('bool').astype('int')
```

**Union con el Dataset de Aeropuertos**

Ahora que la limpieza esta terminada paso a unirlo con el otro dataset secundario descargado. Para esto tenia que tener una llave que pueda utilizarse para conbinar ambos datasets como es el IATA Code.

```python
import pandas as pd
airports_path = "raw data/airports.csv" 
airports = pd.read_csv(airports_path, dtype=str)

airports = airports.rename(columns={col: col.strip() for col in airports.columns})
airports = airports[['iata_code','name','latitude_deg','longitude_deg','iso_country']]
# Normalizar iata_code y convertir lat/lon a numerico
airports['iata_code'] = airports['iata_code'].astype(str).str.strip().str.upper()
airports['latitude_deg'] = pd.to_numeric(airports['latitude_deg'], errors='coerce')
airports['longitude_deg'] = pd.to_numeric(airports['longitude_deg'], errors='coerce')


# Merge para origen (left join)
df_merged = df_clean.merge(
    airports[['iata_code','latitude_deg','longitude_deg']].rename(
        columns={'iata_code':'Origin_iata','latitude_deg':'origin_lat','longitude_deg':'origin_lon'}
    ),
    how='left',
    left_on='Origin',
    right_on='Origin_iata'
).drop(columns=['Origin_iata'])

# Merge para destino (left join)
df_merged = df_merged.merge(
    airports[['iata_code','latitude_deg','longitude_deg']].rename(
        columns={'iata_code':'Dest_iata','latitude_deg':'dest_lat','longitude_deg':'dest_lon'}
    ),
    how='left',
    left_on='Dest',
    right_on='Dest_iata'
).drop(columns=['Dest_iata'])

# Comprobaciones y reporte rapido
missing_origin = df_merged['origin_lat'].isna().sum()
missing_dest = df_merged['dest_lat'].isna().sum()
total = len(df_merged)
print(f"Faltan coords origen: {missing_origin} / {total}")
print(f"Faltan coords destino: {missing_dest} / {total}")

# Opcional: listar codigos IATA sin match (para inspeccion)
missing_orig_codes = (df_merged[df_merged['origin_lat'].isna()]['Origin']
                      .drop_duplicates().sort_values().tolist())
missing_dest_codes = (df_merged[df_merged['dest_lat'].isna()]['Dest']
                      .drop_duplicates().sort_values().tolist())
print("Origin sin match (ejemplos):", missing_orig_codes[:20])
print("Dest sin match (ejemplos):", missing_dest_codes[:20])

```

```python
import re
import pandas as pd

def normalize_tail(s):
    if pd.isna(s): return s
    s = str(s).strip().upper()
    s = re.sub(r'[^A-Z0-9]', '', s)
    if s == '': return s
    if s[0].isdigit(): s = 'N' + s
    if s.startswith('NN'): s = s[1:]
    return s

df = df_clean.copy()
faa = pd.read_csv("raw data/faa_tailnumbers_filtered.csv", dtype=str, low_memory=False)

# Normalizar matrículas en df
df['Tail_Number_norm'] = df['Tail_Number'].astype(str).apply(normalize_tail)

# Detectar columna matrícula en FAA y normalizar
orig_cols = list(faa.columns)
cols_map = {c.lower().replace(' ','').replace('-','').replace('_',''): c for c in orig_cols}
candidates_n = ['n-number','n_number','nnumber','registration','registrationnumber','nnum','n']
col_n = next((cols_map[c] for c in candidates_n if c in cols_map), None)
if col_n is None:
    for c in orig_cols:
        sample = faa[c].dropna().astype(str).head(50).tolist()
        if sample and sum(bool(re.search(r'[A-Z]', v) and re.search(r'\d', v)) for v in sample) >= 3:
            col_n = c
            break
if col_n is None:
    col_n = orig_cols[0]

faa['N_NUMBER_norm'] = faa[col_n].astype(str).apply(normalize_tail)

# Detectar columnas model y year en FAA
col_model = next((cols_map[c] for c in ['model','mfr_mdl_code','maker','modelname','aircraftmodel','mfr_model','mfrmdlcode'] if c in cols_map), None)
col_year  = next((cols_map[c] for c in ['yearmfr','year','year_mfr','manufactureyear','yearmanufactured','yearmfg'] if c in cols_map), None)

faa['MODEL'] = faa[col_model].astype(str).str.strip() if col_model else pd.NA
faa['YEAR_MFR'] = faa[col_year].astype(str).str.strip() if col_year else pd.NA

# Merge por matrícula normalizada
faa_for_merge = faa[['N_NUMBER_norm','MODEL','YEAR_MFR']].rename(columns={'N_NUMBER_norm':'N_NUMBER'})
df = df.merge(faa_for_merge, how='left', left_on='Tail_Number_norm', right_on='N_NUMBER')

# Calcular aircraft_age
df['FlightDate'] = pd.to_datetime(df.get('FlightDate', pd.Series()), errors='coerce')
df['YEAR_MFR'] = pd.to_numeric(df['YEAR_MFR'], errors='coerce')
df['aircraft_age'] = df['FlightDate'].dt.year - df['YEAR_MFR']

# Cobertura compacta
total = len(df)
from_faa = df['MODEL'].notna().sum()
print(f"Total filas: {total}")
print(f"Coincidencias con FAA (modelo encontrado): {from_faa}")
print(f"Sin match (modelo NaN): {total - from_faa}")
print(f"Porcentaje de cobertura: {from_faa/total:.2%}")

# Inferir modelos faltantes por Operating_Airline (fallback global si hace falta)
air_col = next((c for c in df.columns if c.lower() in ('operating_airline','operator','airline','operatingcarrier')), None)
df['Operating_Airline'] = df[air_col].astype(str).fillna('UNKNOWN') if air_col else 'UNKNOWN'

mode_by_airline = (df[df['MODEL'].notna()]
                   .groupby('Operating_Airline')['MODEL']
                   .agg(lambda s: s.mode().iat[0] if not s.mode().empty else pd.NA)
                   .to_dict())

global_mode = df['MODEL'].dropna().mode()
global_mode = global_mode.iat[0] if not global_mode.empty else pd.NA

def infer_model(row):
    if pd.notna(row['MODEL']) and str(row['MODEL']).lower() != 'nan': return row['MODEL']
    a = row['Operating_Airline']
    if a in mode_by_airline and pd.notna(mode_by_airline[a]): return mode_by_airline[a]
    return global_mode

df['MODEL_final'] = df.apply(infer_model, axis=1)
df['MODEL_source'] = df.apply(lambda r: 'faa' if pd.notna(r['MODEL']) and str(r['MODEL']).lower()!='nan' else ('inferred' if pd.notna(r['MODEL_final']) else 'missing'), axis=1)

# Resultado final: solo las columnas solicitadas
out = df[['Tail_Number_norm','N_NUMBER','MODEL','YEAR_MFR','aircraft_age','MODEL_final','MODEL_source']].copy()

from_faa = (out['MODEL_source']=='faa').sum()
inferred = (out['MODEL_source']=='inferred').sum()
final_coverage = (from_faa + inferred) / total if total else 0
print(f"Model desde FAA: {from_faa} ({from_faa/total:.2%})")
print(f"Model inferidos: {inferred} ({inferred/total:.2%})")
print(f"Cobertura final (FAA + inferidos): {final_coverage:.2%}")
```
