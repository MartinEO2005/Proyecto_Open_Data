# Extraccion del Archivo de Kaggle

**Puede tardar mucho tiempo**- igualmente el archivo lo puse dentro del directorio raw data.

```python
from kaggle.api.kaggle_api_extended import KaggleApi
import os

# Autenticar con Kaggle
api = KaggleApi()
api.authenticate()

print("✓ Autenticación con Kaggle completada")
```

```python
# Descargar el archivo Combined_Flights_2022.csv desde Kaggle
import zipfile

dataset = 'robikscube/flight-delay-dataset-20182022'
file_name = 'Combined_Flights_2021.csv'

# Obtener la ruta actual del notebook
output_path = os.getcwd()

# Descargar el archivo específico
api.dataset_download_file(dataset, file_name, path=output_path)

print(f"✓ Descarga completada\n")

# El archivo se descarga como ZIP, vamos a descomprimirlo
zip_file = os.path.join(output_path, f'{file_name}.zip')

if os.path.exists(zip_file):
    print(f"Descomprimiendo {os.path.basename(zip_file)}...")
    with zipfile.ZipFile(zip_file, 'r') as zip_ref:
        zip_ref.extractall(output_path)
    
    # Eliminar el archivo ZIP después de descomprimir
    os.remove(zip_file)
    print(f"✓ Archivo descomprimido y ZIP eliminado\n")

# Verificar que el archivo CSV existe
csv_path = os.path.join(output_path, file_name)
if os.path.exists(csv_path):
    file_size = os.path.getsize(csv_path)
    print(f"✓ Archivo {file_name} disponible")
    print(f"✓ Tamaño: {file_size / (1024**2):.2f} MB")
else:
    print(f"✗ Error: El archivo {file_name} no se encontró")
```

```python
import hangarbay as hb
import pandas as pd

# 1) Cargar y preparar la base FAA (solo la primera vez tarda más)
hb.load_data()

# 2) Leer tu CSV de tail numbers (una columna, sin encabezado)
tails = pd.read_csv("raw data/tail_numbers.csv", header=None, names=["Tail_Number"], dtype=str)

# 3) Normalizar tail numbers: string, trim, upper, quitar caracteres raros
tails['Tail_Number'] = (
    tails['Tail_Number']
    .astype(str)
    .str.strip()
    .str.upper()
    .str.replace(r'[^A-Z0-9]', '', regex=True)
)

# 4) Usar valores únicos para evitar búsquedas repetidas
unique_tails = tails['Tail_Number'].dropna().unique()

# 5) Buscar en FAA con hangarbay, acumulando resultados no vacíos
frames = []
for tn in unique_tails:
    try:
        df = hb.search(tn)
        if isinstance(df, pd.DataFrame) and not df.empty:
            frames.append(df)
    except Exception:
        # Si alguna búsqueda falla, la saltamos para que el proceso sea robusto
        pass

# 6) Concatenar todo y seleccionar columnas relevantes
if frames:
    results = pd.concat(frames, ignore_index=True)
else:
    results = pd.DataFrame(columns=['N-NUMBER','MODEL','YEAR MFR'])

# 7) Guardar el filtrado de FAA actualizado
results.to_csv("raw data/faa_tailnumbers_filtered.csv", index=False)

print("Filtrados:", results.shape[0], "registros guardados en faa_tailnumbers_filtered.csv")
print(results[['N-NUMBER','MODEL','YEAR MFR']].head())
```
