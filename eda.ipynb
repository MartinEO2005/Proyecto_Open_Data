0. **grafica para Porcentaje del dataset que fue Cancelled o Diverted (No limpio)**

```python
import matplotlib.pyplot as plt
import pandas as pd

cond_cancelled = df_no_limpio['Cancelled'].fillna(0).astype(int) == 1
cond_diverted = (~cond_cancelled) & (df_no_limpio['Diverted'].fillna(0).astype(int) == 1)
cond_delayed = (~cond_cancelled) & (~cond_diverted) & ( (df_no_limpio['DepDelayMinutes'] > 15) | (df_no_limpio['ArrDelayMinutes'] > 15) )
cond_on_time = (~cond_cancelled) & (~cond_diverted) & (~(df_no_limpio['DepDelayMinutes'] > 15)) & (~(df_no_limpio['ArrDelayMinutes'] > 15))

sizes = [
    int(cond_on_time.sum()),
    int(cond_delayed.sum()),
    int(cond_cancelled.sum()),
    int(cond_diverted.sum())
]
labels = ['On-time (<=15 min total)', 'Delayed (>15 min total)', 'Cancelled', 'Diverted']
colors = ['#2ca02c','#ff7f0e','#d62728','#9467bd']
plt.figure(figsize=(8, 7))
patches, texts, autotexts = plt.pie(sizes, 
                                    colors=colors, 
                                    autopct='%1.1f%%', 
                                    startangle=90, 
                                    pctdistance=0.75) 

centre = plt.Circle((0,0),0.50,fc='white')
plt.gcf().gca().add_artist(centre)
plt.title('Grafica 0: Composición global de vuelos en el Dataset Original')
autotexts[2].set_position((0.1, 0.4))
autotexts[3].set_position((-0.1, 0.4))

plt.legend(labels, loc='center left', bbox_to_anchor=(1, 0.5))
plt.tight_layout()
plt.show()
```

1. **grafica para Porcentaje del dataset que tuvo delay (>15 min)**

```python
total = len(df)
n_delayed_total = int((df['TotalDelayMinutes'] > 15).sum())
pct_delayed_total = 100 * n_delayed_total / total

print(f"Total filas: {total:,}")
print(f"Vuelos delayed (TotalDelayMinutes > 15): {n_delayed_total:,} ({pct_delayed_total:.2f}%)")

plot_df = pd.DataFrame({
    "Estado": ["On-time (<=15 min total)", "Delayed (>15 min total)"],
    "Vuelos": [total - n_delayed_total, n_delayed_total]
})

plt.figure(figsize=(7,4))
sns.barplot(data=plot_df, x="Estado", y="Vuelos", hue="Estado",
            palette={"On-time (<=15 min total)":"#2ca02c", "Delayed (>15 min total)":"#ff7f0e"},
            dodge=False)
plt.title('Grafica 1:On-time vs Delayed (TotalDelayMinutes > 15)')
plt.ylabel('Número de vuelos')
for i, row in plot_df.iterrows():
    plt.text(i, 
        row["Vuelos"] + total*0.0025,
        f"{100*row['Vuelos']/total:.2f}%", 
        ha='center',
        va='bottom', 
    )
plt.legend([],[], frameon=False)
plt.tight_layout()
plt.show()
```

7. **Modelos por Familia**

- **7.1 y 7.2 graficas por familia**
- **Graficas individuales por familia**

```python
import seaborn as sns
import matplotlib.pyplot as plt

def model_family(model):
    if isinstance(model, float):  # NaN
        return 'Unknown'
    if model.startswith('737'):
        return 'B737'
    if model.startswith('757'):
        return 'B757'
    if model.startswith('767'):
        return 'B767'
    if model.startswith('777'):
        return 'B777'
    if model.startswith('787'):
        return 'B787'
    if model.startswith(('A319', 'A320', 'A321')):
        return 'A320-family'
    if model.startswith('A330'):
        return 'A330'
    if model.startswith('A350'):
        return 'A350'
    if model.startswith(('EMB', 'ERJ')):
        return 'Embraer'
    if model.startswith('CL-600'):
        return 'CRJ'
    if model.startswith('DHC'):
        return 'Turboprop'
    return 'Other'
agg_family = (
    df.groupby(df['MODEL'].apply(model_family))
      .agg(
          n_flights=('FlightDate', 'count'),
          mean_arr_delay=('ArrDelayMinutes', 'mean')
      )
      .reset_index()
      .rename(columns={'MODEL': 'MODEL_FAMILY'})
)

agg_family_sorted = agg_family.sort_values('mean_arr_delay', ascending=False)

df['delay15'] = df['ArrDelayMinutes'] > 15

agg_family_otp = (
    df.groupby(df['MODEL'].apply(model_family))
      .agg(
          n_flights=('FlightDate', 'count'),
          pct_delay15=('delay15', 'mean')
      )
      .reset_index()
      .rename(columns={'MODEL': 'MODEL_FAMILY'})
)

agg_family_otp['pct_delay15'] *= 100
agg_family_otp = agg_family_otp.sort_values('pct_delay15', ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(18, 8))

sns.barplot(
    data=agg_family_sorted,
    x='mean_arr_delay',
    y='MODEL_FAMILY',
    palette='Greens_r',
    ax=axes[0]
)
axes[0].set_title('7.1 Retraso medio por familia de modelos')
axes[0].set_xlabel('Retraso medio (min)')
axes[0].set_ylabel('Familia de modelos')

sns.barplot(
    data=agg_family_otp,
    x='pct_delay15',
    y='MODEL_FAMILY',
    palette='Reds_r',
    ax=axes[1]
)
axes[1].set_title('7.2 % de vuelos con retraso >15 min por familia')
axes[1].set_xlabel('% de vuelos retrasados')
axes[1].set_ylabel('Familia de modelos')

plt.tight_layout()
plt.show()
```

```python
import math
families = df['MODEL'].apply(model_family).unique()
n_fam = len(families)
#   CONFIGURAR GRID DE SUBPLOTS
cols = 3
rows = math.ceil(n_fam / cols)

fig, axes = plt.subplots(rows, cols, figsize=(20, rows * 5))
axes = axes.flatten()

for i, fam in enumerate(families):
    ax = axes[i]

    # Filtrar vuelos de esa familia
    df_fam = df[df['MODEL'].apply(model_family) == fam]

    # Agregar por modelo dentro de la familia
    agg_models = (
        df_fam.groupby('MODEL')
              .agg(
                  n_flights=('FlightDate', 'count'),
                  mean_arr_delay=('ArrDelayMinutes', 'mean')
              )
              .reset_index()
              .sort_values('mean_arr_delay', ascending=False)
    )
    sns.barplot(
        data=agg_models,
        x='mean_arr_delay',
        y='MODEL',
        palette='viridis',
        ax=ax
    )

    ax.set_title(f'Familia {fam}')
    ax.set_xlabel('Retraso medio (min)')
    ax.set_ylabel('Modelo')

# Ocultar subplots vacíos si sobran
for j in range(i+1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

## **Rama 3:** Impacto de congestion y Franjas horarias

10. **Distribucion por mes/ o dia en categorias**

12. **Lag Plots**

```python
delay_rate = (
    df.groupby('Origin')['DepDelayMinutes']
      .apply(lambda x: (x > 15).mean())
      .sort_values(ascending=False)
)
top_delay_airports = delay_rate.head(5).index
low_delay_airports = delay_rate.tail(5).index

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Configuración de estilo
sns.set_theme(style="whitegrid")

def plot_improved_lags(airports, title_suffix):
    fig, axes = plt.subplots(1, 5, figsize=(25, 6), sharey=True, sharex=True)
    
    for i, ap in enumerate(airports):
        df_ap = df[df['Origin'] == ap].sort_values('dep_dt').copy()
        df_ap['lag_delay'] = df_ap['DepDelayMinutes'].shift(1)
        
        # Eliminamos NaNs para el cálculo de correlación
        valid_data = df_ap.dropna(subset=['lag_delay', 'DepDelayMinutes'])
        corr = valid_data['lag_delay'].corr(valid_data['DepDelayMinutes'])

        # Usamos regplot para ver la línea de tendencia
        sns.regplot(
            data=valid_data,
            x='lag_delay',
            y='DepDelayMinutes',
            ax=axes[i],
            scatter_kws={'alpha': 0.15, 's': 10, 'color': '#2c3e50'},
            line_kws={'color': 'red', 'lw': 1.5}
        )

        # Añadimos línea y = x para referencia de persistencia
        axes[i].plot([0, 500], [0, 500], ls="--", c=".3", alpha=0.5)
        
        axes[i].set_title(f'12.1 {ap}\nCorr: {corr:.2f}', fontsize=12, fontweight='bold')
        axes[i].set_xlabel('Retraso (t-1)')
        axes[i].set_ylabel('Retraso (t)' if i == 0 else '')
        
        # Limitamos para ver mejor la masa de datos
        axes[i].set_xlim(-5, 300)
        axes[i].set_ylim(-5, 300)

    plt.suptitle(f'Análisis de Persistencia de Retrasos - {title_suffix}', fontsize=16, y=1.05)
    plt.tight_layout()
    plt.show()
# Ejecutar para ambos casos
plot_improved_lags(top_delay_airports, "Alta Incidencia")
plot_improved_lags(low_delay_airports, "Baja Incidencia")
```

```python
from scipy.stats import pearsonr
from sklearn.linear_model import LinearRegression

df['dep_dt'] = pd.to_datetime(df['dep_dt'], errors='coerce')
df['arr_dt'] = pd.to_datetime(df['arr_dt'], errors='coerce')
df_dep = df[df['TaxiOut'].notna() & df['DepDelayMinutes'].notna() & (df['TaxiOut'] >= 0)]
df_arr = df[df['TaxiIn'].notna() & df['ArrDelayMinutes'].notna() & (df['TaxiIn'] >= 0)]

# TaxiOut → DepDelay
X_dep = df_dep['TaxiOut'].values.reshape(-1,1)
y_dep = df_dep['DepDelayMinutes'].values
model_dep = LinearRegression().fit(X_dep, y_dep)
r_dep, _ = pearsonr(df_dep['TaxiOut'], df_dep['DepDelayMinutes'])

# TaxiIn → ArrDelay
X_arr = df_arr['TaxiIn'].values.reshape(-1,1)
y_arr = df_arr['ArrDelayMinutes'].values
model_arr = LinearRegression().fit(X_arr, y_arr)
r_arr, _ = pearsonr(df_arr['TaxiIn'], df_arr['ArrDelayMinutes'])

fig, axes = plt.subplots(1, 2, figsize=(14,6))
# --- Subplot 1: TaxiOut ---
sns.scatterplot(data=df_dep, x='TaxiOut', y='DepDelayMinutes', alpha=0.25, s=10, ax=axes[0])
x_vals = np.linspace(0, df_dep['TaxiOut'].max(), 200)
axes[0].plot(x_vals, model_dep.predict(x_vals.reshape(-1,1)), color='red', label=f'Pearson r={r_dep:.3f}')
axes[0].set_title('13.1 TaxiOut vs DepDelay')
axes[0].set_xlabel('TaxiOut (min)')
axes[0].set_ylabel('DepDelayMinutes (min)')
axes[0].legend(loc='upper left')

# --- Subplot 2: TaxiIn ---
sns.scatterplot(data=df_arr, x='TaxiIn', y='ArrDelayMinutes', alpha=0.25, s=10, ax=axes[1])
x_vals2 = np.linspace(0, df_arr['TaxiIn'].max(), 200)
axes[1].plot(x_vals2, model_arr.predict(x_vals2.reshape(-1,1)), color='red', label=f'Pearson r={r_arr:.3f}')
axes[1].set_title('13.2 TaxiIn vs ArrDelay')
axes[1].set_xlabel('TaxiIn (min)')
axes[1].set_ylabel('ArrDelayMinutes (min)')
axes[1].legend(loc='upper left')

plt.tight_layout()
plt.show()
```

```python
# Mapa interactivo por MES (ruta única, color = retraso medio, grosor = volumen)
import pandas as pd
import numpy as np
import folium
from branca.colormap import linear
import ipywidgets as widgets
from IPython.display import display

# ---------- 2) Preparar columnas ----------
df['FlightDate'] = pd.to_datetime(df['FlightDate'], errors='coerce')
df['month'] = df['FlightDate'].dt.month

# Asegurar columnas numéricas relevantes
df['Distance'] = pd.to_numeric(df['Distance'], errors='coerce')
df['ArrDelayMinutes'] = pd.to_numeric(df['ArrDelayMinutes'], errors='coerce')

# ---------- 3) Función que genera y guarda el mapa para un mes dado ----------
def generar_mapa_por_mes(mes, min_distance=1000):
    # Filtrar por mes y por distancia mínima
    df_f = df[df['month'] == mes].copy()
    df_f = df_f[df_f['Distance'] >= min_distance]
    df_f = df_f.dropna(subset=['origin_lat','origin_lon','dest_lat','dest_lon','ArrDelayMinutes'])
    
    if df_f.empty:
        print(f"No hay datos para mes={mes} con Distance >= {min_distance}")
        return None

    # Agregar rutas únicas y calcular métricas por ruta
    route_stats = (
        df_f.groupby(['Origin','Dest','origin_lat','origin_lon','dest_lat','dest_lon'])
            .agg(
                vol=('ArrDelayMinutes','size'),
                avg_delay=('ArrDelayMinutes','mean'),
                distance=('Distance','mean')
            )
            .reset_index()
    )

    # Colormap continuo sobre retraso medio
    colormap = linear.YlOrRd_09.scale(route_stats['avg_delay'].min(), route_stats['avg_delay'].max())

    # Crear mapa centrado en EEUU
    m = folium.Map(location=[39.5, -98.35], zoom_start=4, tiles='CartoDB positron')

    # Dibujar rutas: color por retraso medio, grosor por volumen
    for _, r in route_stats.iterrows():
        color = colormap(r['avg_delay'])
        weight = 1 + np.log1p(r['vol'])
        folium.PolyLine(
            locations=[(r['origin_lat'], r['origin_lon']), (r['dest_lat'], r['dest_lon'])],
            color=color,
            weight=weight,
            opacity=0.8,
            tooltip=(
                f"{r['Origin']}→{r['Dest']} | Dist: {r['distance']:.0f} km | "
                f"Vuelos: {r['vol']} | Retraso medio: {r['avg_delay']:.1f} min"
            )
        ).add_to(m)

    # Añadir barra de color
    colormap.caption = "Retraso medio (min)"
    colormap.add_to(m)

    # Guardar mapa en HTML
    filename = f"mapa_rutas_mes_{mes:02d}.html"
    m.save(filename)
    print(f"Mapa guardado en: {filename}")
    return filename

# ---------- 4) Widgets: selector de mes y distancia mínima ----------
meses_disponibles = sorted(df['month'].dropna().unique().astype(int).tolist())
mes_dropdown = widgets.Dropdown(options=meses_disponibles, description='Mes:')
dist_slider = widgets.IntSlider(value=1000, min=100, max=5000, step=100, description='Dist mín (millas):')
boton = widgets.Button(description='Generar mapa (guardar HTML)')
out = widgets.Output()

def on_click(b):
    with out:
        out.clear_output()
        mes = int(mes_dropdown.value)
        min_dist = int(dist_slider.value)
        generar_mapa_por_mes(mes, min_distance=min_dist)

boton.on_click(on_click)

display(mes_dropdown, dist_slider, boton, out)
```

```python
import pandas as pd
import numpy as np
import folium
from branca.colormap import linear
import ipywidgets as widgets
from IPython.display import display


# ---------- 2) Preparar columnas de fecha ----------
df['FlightDate'] = pd.to_datetime(df['FlightDate'], errors='coerce')
df['month'] = df['FlightDate'].dt.month
df['day'] = df['FlightDate'].dt.day

# Asegurar columnas numéricas
df['Distance'] = pd.to_numeric(df['Distance'], errors='coerce')
df['ArrDelayMinutes'] = pd.to_numeric(df['ArrDelayMinutes'], errors='coerce')

# ---------- 3) Función para generar mapa por rango ----------
def generar_mapa_rango(mes, dia_inicio, dia_fin, min_distance=1000):
    df_f = df[
        (df['month'] == mes) &
        (df['day'] >= dia_inicio) &
        (df['day'] <= dia_fin) &
        (df['Distance'] >= min_distance)   # ahora en millas
    ].dropna(subset=['origin_lat','origin_lon','dest_lat','dest_lon','ArrDelayMinutes'])

    if df_f.empty:
        print(f"No hay datos para mes {mes}, días {dia_inicio}–{dia_fin}")
        return None

    # Agregar rutas
    route_stats = (
        df_f.groupby(['Origin','Dest','origin_lat','origin_lon','dest_lat','dest_lon'])
            .agg(
                vol=('ArrDelayMinutes','size'),
                avg_delay=('ArrDelayMinutes','mean'),
                distance=('Distance','mean')   # millas
            )
            .reset_index()
    )

    # Colormap
    colormap = linear.YlOrRd_09.scale(route_stats['avg_delay'].min(), route_stats['avg_delay'].max())

    # Crear mapa
    m = folium.Map(location=[39.5, -98.35], zoom_start=4, tiles='CartoDB positron')

    for _, r in route_stats.iterrows():
        folium.PolyLine(
            locations=[(r['origin_lat'], r['origin_lon']), (r['dest_lat'], r['dest_lon'])],
            color=colormap(r['avg_delay']),
            weight=1 + np.log1p(r['vol']),
            opacity=0.8,
            tooltip=(
                f"{r['Origin']}→{r['Dest']}<br>"
                f"Distancia: {r['distance']:.0f} mi<br>"   # ✅ millas
                f"Vuelos: {r['vol']}<br>"
                f"Retraso medio: {r['avg_delay']:.1f} min"
            )
        ).add_to(m)

    colormap.caption = "Retraso medio (min)"
    colormap.add_to(m)

    # Guardar
    filename = f"mapa_rutas_mes{mes:02d}_dias{dia_inicio:02d}-{dia_fin:02d}.html"
    m.save(filename)
    print(f"Mapa guardado en: {filename}")
    return filename

# ---------- 4) Widgets interactivos ----------
meses = sorted(df['month'].dropna().unique().astype(int).tolist())
dias = list(range(1,32))

mes_dropdown = widgets.Dropdown(options=meses, description='Mes:')
dia_inicio_dropdown = widgets.Dropdown(options=dias, description='Día inicio:')
dia_fin_dropdown = widgets.Dropdown(options=dias, description='Día fin:')
dist_slider = widgets.IntSlider(value=1000, min=100, max=5000, step=100, description='Dist mín (mi):')  # ✅ millas
boton = widgets.Button(description='Generar mapa')
out = widgets.Output()

def on_click(b):
    with out:
        out.clear_output()
        mes = int(mes_dropdown.value)
        d1 = int(dia_inicio_dropdown.value)
        d2 = int(dia_fin_dropdown.value)
        dist = int(dist_slider.value)
        generar_mapa_rango(mes, d1, d2, min_distance=dist)

boton.on_click(on_click)
display(mes_dropdown, dia_inicio_dropdown, dia_fin_dropdown, dist_slider, boton, out)
```

```python
#   MAPA POR MODELO (familia → modelo)
#   Filtrar por uno, varios o TODOS los meses + rango de días
#   Color = retraso medio | Grosor = volumen

import pandas as pd
import numpy as np
import folium
from branca.colormap import linear
import ipywidgets as widgets
from IPython.display import display


df['FlightDate'] = pd.to_datetime(df['FlightDate'], errors='coerce')
df['month'] = df['FlightDate'].dt.month
df['day'] = df['FlightDate'].dt.day

df['Distance'] = pd.to_numeric(df['Distance'], errors='coerce')
df['ArrDelayMinutes'] = pd.to_numeric(df['ArrDelayMinutes'], errors='coerce')


def model_family(model):
    if isinstance(model, float):  # NaN
        return 'Unknown'
    if model.startswith('737'):
        return 'B737'
    if model.startswith('757'):
        return 'B757'
    if model.startswith('767'):
        return 'B767'
    if model.startswith('777'):
        return 'B777'
    if model.startswith('787'):
        return 'B787'
    if model.startswith(('A319', 'A320', 'A321')):
        return 'A320-family'
    if model.startswith('A330'):
        return 'A330'
    if model.startswith('A350'):
        return 'A350'
    if model.startswith(('EMB', 'ERJ')):
        return 'Embraer'
    if model.startswith('CL-600'):
        return 'CRJ'
    if model.startswith('DHC'):
        return 'Turboprop'
    return 'Other'

df['model_family'] = df['MODEL'].apply(model_family)


def generar_mapa_modelo(familia, modelo, meses, dia_inicio, dia_fin, min_distance=1000):

    if "Todos" in meses:
        meses = sorted(df['month'].dropna().unique().astype(int).tolist())

    df_f = df[
        (df['model_family'] == familia) &
        (df['MODEL'] == modelo) &
        (df['month'].isin(meses)) &
        (df['day'] >= dia_inicio) &
        (df['day'] <= dia_fin) &
        (df['Distance'] >= min_distance)
    ].dropna(subset=['origin_lat','origin_lon','dest_lat','dest_lon','ArrDelayMinutes'])

    if df_f.empty:
        print("No hay datos para ese filtro.")
        return None

    # Agregar rutas
    route_stats = (
        df_f.groupby(['Origin','Dest','origin_lat','origin_lon','dest_lat','dest_lon'])
            .agg(
                vol=('ArrDelayMinutes','size'),
                avg_delay=('ArrDelayMinutes','mean'),
                distance=('Distance','mean')
            )
            .reset_index()
    )

    colormap = linear.YlOrRd_09.scale(
        route_stats['avg_delay'].min(),
        route_stats['avg_delay'].max()
    )

    m = folium.Map(location=[39.5, -98.35], zoom_start=4, tiles='CartoDB positron')

    # Dibujar rutas
    for _, r in route_stats.iterrows():
        folium.PolyLine(
            locations=[(r['origin_lat'], r['origin_lon']),
                       (r['dest_lat'], r['dest_lon'])],
            color=colormap(r['avg_delay']),
            weight=1 + np.log1p(r['vol']),
            opacity=0.8,
            tooltip=(
                f"{r['Origin']}→{r['Dest']}<br>"
                f"Modelo: {modelo}<br>"
                f"Distancia: {r['distance']:.0f} mi<br>"
                f"Vuelos: {r['vol']}<br>"
                f"Retraso medio: {r['avg_delay']:.1f} min"
            )
        ).add_to(m)

    colormap.caption = "Retraso medio (min)"
    colormap.add_to(m)

    # Guardar archivo
    meses_str = "-".join(str(m) for m in meses)
    filename = f"mapa_modelo_{familia}_{modelo}_meses_{meses_str}_dias{dia_inicio}-{dia_fin}.html"
    m.save(filename)
    print(f"Mapa guardado en: {filename}")
    return filename


# Familias disponibles
familias = sorted(df['model_family'].dropna().unique().tolist())
familia_dropdown = widgets.Dropdown(options=familias, description='Familia:')

def actualizar_modelos(*args):
    fam = familia_dropdown.value
    modelos = sorted(df[df['model_family']==fam]['MODEL'].dropna().unique().tolist())
    modelo_dropdown.options = modelos

modelo_dropdown = widgets.Dropdown(description='Modelo:')
familia_dropdown.observe(actualizar_modelos, names='value')
actualizar_modelos()

# Meses (multi-select)
meses_disponibles = sorted(df['month'].dropna().unique().astype(int).tolist())
meses_dropdown = widgets.SelectMultiple(
    options=["Todos"] + meses_disponibles,
    description='Mes(es):',
    rows=6
)

# Días
dias = list(range(1,32))
dia_inicio_dropdown = widgets.Dropdown(options=dias, description='Día inicio:')
dia_fin_dropdown = widgets.Dropdown(options=dias, description='Día fin:')

# Distancia mínima
dist_slider = widgets.IntSlider(value=1000, min=100, max=5000, step=100, description='Dist mín (mi):')

# Botón
boton = widgets.Button(description='Generar mapa')
out = widgets.Output()

def on_click(b):
    with out:
        out.clear_output()
        generar_mapa_modelo(
            familia_dropdown.value,
            modelo_dropdown.value,
            list(meses_dropdown.value),
            int(dia_inicio_dropdown.value),
            int(dia_fin_dropdown.value),
            min_distance=int(dist_slider.value)
        )

boton.on_click(on_click)

display(familia_dropdown, modelo_dropdown, meses_dropdown, dia_inicio_dropdown, dia_fin_dropdown, dist_slider, boton, out)
```
