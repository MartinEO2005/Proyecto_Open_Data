```python
from graphviz import Digraph

dot = Digraph(comment='ER Diagram Normalized')

# Entidades
dot.node('AircraftModel', 'AircraftModel')
dot.node('Aircraft', 'Aircraft')
dot.node('Airline', 'Airline')
dot.node('Airport', 'Airport')
dot.node('Flight', 'Flight')
dot.node('FlightPerformance', 'FlightPerformance')

# Relaciones
dot.edge('AircraftModel', 'Aircraft', label='1:N')
dot.edge('Aircraft', 'Flight', label='1:N')
dot.edge('Airline', 'Flight', label='1:N')
dot.edge('Airport', 'Flight', label='Origin/Dest')
dot.edge('Flight', 'FlightPerformance', label='1:1')

# Renderizar a PNG
dot.render('er_diagram_normalized', format='png', cleanup=True)
print("Diagrama generado: er_diagram_normalized.png")
```
